## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

Q1 ### Présentation
- **Nom & Famille :** Régression linéaire / *Linear Regression*. Famille de la **régression** (apprentissage supervisé). Modéliser la relation entre une variable cible continue $y$ et des variables explicatives $X$.
- **Intuition et étapes :** Recherche la meilleure droite (ou hyperplan) pour minimiser l'écart quadratique moyen.
  - *Entrée :* Données numériques $X$ et vecteur $y$.
  - *Apprentissage :* Estimation des coefficients par les moindres carrés ordinaires (MCO).
  - *Sortie :* Valeur numérique continue prédite $\hat{y}$.
- **Formule mathématique :**
  $$\hat{y} = \beta_0 + \sum_{j=1}^{p} \beta_j X_j$$
  - *Symboles :* $\hat{y}$ est la prédiction, $X_j$ la variable $j$, $\beta_0$ le biais, $\beta_j$ les coefficients de pente.
- **Avantages et limites :**
  - *Avantages :* Très simple à interpréter et coût de calcul minimal.
  - *Limites :* Incompatible avec les relations non linéaires et sensible aux valeurs aberrantes.

Q2### Schéma explicatif
Principe de l'algorithme en 3 étapes :
1. **Entrée :** Données numériques ($X, y$)
2. **Calcul / Apprentissage :** Optimisation par minimisation des erreurs (moindres carrés)
3. **Sortie :** Prédiction continue ($\hat{y}$)
*Commentaire de lecture :* Le modèle ajuste une droite au nuage de points pour minimiser la distance verticale (résidus).

Q3### Contextes d'utilisation
- **Cas d'usage concrets :**
  1. *Finance/Gestion :* Prédiction du chiffre d'affaires annuel selon les dépenses marketing.
  2. *Immobilier :* Estimation du prix de vente d'un bien selon sa superficie.
- **Nature des données & Prérequis :** Variables numériques continues sans colinéarité stricte.
- **Cas peu approprié :** Classification binaire (ex. détection de fraude), car la sortie n'est pas bornée entre 0 et 1.
- **Métriques adaptées :** MAE, RMSE, $R^2$.

Q4### Analyse de l'expérience Python
- **Entrées & Hyperparamètres :** $X_{reg}$ et $y_{reg}$. Aucun hyperparamètre majeur réglé.
- **Sortie obtenue :** La valeur de l'erreur absolue moyenne (MAE).
- **Conclusion :** Mesure la déviation moyenne sur les données test. Ne valide pas la robustesse en environnement économique réel.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

Q1### Présentation
- **Nom & Famille :** Régression Ridge / *Ridge Regression*. Famille de la **régression avec régularisation $L_2$** (apprentissage supervisé). Réduire la variance et éviter le surapprentissage.
- **Intuition et étapes :** Ajoute une pénalité basée sur le carré de la magnitude des coefficients.
  - *Entrée :* Caractéristiques $X$ normalisées et cible $y$.
  - *Apprentissage :* Minimisation de la somme des erreurs quadratiques augmentée de la pénalité $L_2$.
  - *Sortie :* Prédiction continue $\hat{y}$.
- **Formule mathématique :**
  $$\min_{\beta} \|y - X\beta\|_2^2 + \alpha \|\beta\|_2^2$$
  - *Symboles :* $\alpha \ge 0$ est l'hyperparamètre de régularisation, $\beta$ le vecteur des paramètres.
- **Avantages et limites :**
  - *Avantages :* Résout le problème de colinéarité et stabilise les coefficients.
  - *Limites :* Ne supprime aucune variable (conserve tous les coefficients non nuls).

Q2### Schéma explicatif
Principe de la régularisation :
1. **Entrée :** Données $X$ normalisées
2. **Calcul / Apprentissage :** Contrainte sur la taille des coefficients $\beta$ via $\alpha$
3. **Sortie :** Modèle stabilisé sans surapprentissage
*Commentaire de lecture :* Quand $\alpha$ augmente, les coefficients sont contractés vers zéro pour lisser la prédiction.

Q3### Contextes d'utilisation
- **Cas d'usage concrets :**
  1. *Finance :* Modélisation du risque de crédit avec des indicateurs macroéconomiques corrélés.
  2. *Marketing :* Prédiction des ventes avec de multiples métriques d'audience superposées.
- **Nature des données & Prérequis :** Données numériques obligatoirement centrées-réduites.
- **Cas peu approprié :** Recherche de parcimonie (sélection explicite de variables).
- **Métriques adaptées :** MAE, RMSE, $R^2$.

Q4### Analyse de l'expérience Python
- **Entrées & Hyperparamètres :** `StandardScaler()` + `Ridge(alpha=1.0)`.
- **Sortie obtenue :** La métrique MAE calculée sur l'échantillon test.
- **Conclusion :** Démontre l'intégration de la régularisation $L_2$ dans un pipeline scikit-learn sans fuite d'information.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom & Famille :** Lasso / *Lasso Regression*. Famille de la **régression avec régularisation $L_1$** (apprentissage supervisé). Sélectionner automatiquement des variables et éviter le surapprentissage.
- **Intuition et étapes :** Ajoute une pénalité basée sur la somme des valeurs absolues des coefficients, forçant certains à devenir nuls.
  - *Entrée :* Données $X$ normalisées et vecteur $y$.
  - *Apprentissage :* Minimisation de la perte quadratique sous contrainte $L_1$.
  - *Sortie :* Prédiction continue $\hat{y}$ et sous-ensemble de variables conservées.
- **Formule mathématique :**
  $$\min_{\beta} \frac{1}{2n} \Vert{}y - X\beta\Vert{}_2^2 + \alpha \Vert{}\beta\Vert{}_1$$
  - *Symboles :* $\alpha$ est l'hyperparamètre de pénalisation, $\Vert{}\beta\Vert{}_1 = \sum \vert{}\beta_j\vert{}$.
- **Avantages et limites :**
  - *Avantages :* Produit des modèles parcimonieux et très interprétables.
  - *Limites :* En cas de forte corrélation entre variables, Lasso en choisit une seule au hasard.

### Schéma explicatif
Principe de la sélection $L_1$ :
1. **Entrée :** Caractéristiques $X$ normalisées
2. **Calcul / Apprentissage :** Annulation stricte des coefficients non pertinents via la contrainte $L_1$
3. **Sortie :** Modèle simplifié avec sélection de variables
*Commentaire de lecture :* La géométrie de la contrainte $L_1$ force certains coefficients à s'annuler exactement, éliminant les variables inutiles.

### Contextes d'utilisation
- **Cas d'usage concrets :**
  1. *Commerce International :* Identification des déterminants majeurs d'exportation parmi des centaines d'indicateurs économiques.
  2. *Santé/Bioinformatique :* Détection de gènes candidats prédictifs.
- **Nature des données & Prérequis :** Variables numériques centrées-réduites.
- **Cas peu approprié :** Données où toutes les variables apportent une contribution indissociable.
- **Métriques adaptées :** $R^2$, MAE, RMSE.

### Analyse de l'expérience Python
- **Entrées & Hyperparamètres :** `StandardScaler()` + `Lasso(alpha=0.1, max_iter=10000)`.
- **Sortie obtenue :** Le coefficient de détermination $R^2$.
- **Conclusion :** Mesure la qualité d'ajustement globale du modèle pénalisé $L_1$ sur le jeu de test.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom & Famille :** Régression polynomiale / *Polynomial Regression*. Famille de la **régression non linéaire** (apprentissage supervisé). Modéliser des relations non linéaires en élevant les variables à différentes puissances.
- **Intuition et étapes :** Transforme l'espace des caractéristiques en générant des termes polynomiaux (carrés, cubes), puis applique une régression linéaire.
  - *Entrée :* Données numériques $X$.
  - *Apprentissage :* Génération de features polynomiales et ajustement des coefficients.
  - *Sortie :* Prédiction continue $\hat{y}$.
- **Formule mathématique :**
  $$\hat{y} = \beta_0 + \beta_1 X + \beta_2 X^2 + \dots + \beta_d X^d$$
  - *Symboles :* $d$ est le degré du polynôme (hyperparamètre), $\beta_j$ les coefficients.
- **Avantages et limites :**
  - *Avantages :* Capte des courbures complexes tout en conservant le cadre linéaire.
  - *Limites :* Explosion dimensionnelle rapide et risque de surapprentissage aux extrémités si le degré est trop élevé.

### Schéma explicatif
Principe de transformation :
1. **Entrée :** Variable simple $X$
2. **Calcul / Apprentissage :** Génération de puissances ($X^2, X^3$) et régression
3. **Sortie :** Courbe polynomiale ajustée
*Commentaire de lecture :* Permet d'épouser les courbures du nuage de points, à condition d'éviter un degré $d$ excessif.

### Contextes d'utilisation
- **Cas d'usage concrets :**
  1. *Logistique et Production :* Modélisation de la consommation d'énergie d'une flotte en fonction de la vitesse.
  2. *Finance :* Estimation de courbes de rendement non linéaires.
- **Nature des données & Prérequis :** Variables numériques continues normalisées.
- **Cas peu approprié :** Jeux de données avec un très grand nombre de variables initiales.
- **Métriques adaptées :** MAE, RMSE, $R^2$.

### Analyse de l'expérience Python
- **Entrées & Hyperparamètres :** `PolynomialFeatures(degree=2)` + `Ridge(alpha=10)`.
- **Sortie obtenue :** La métrique MAE.
- **Conclusion :** Chiffre l'écart moyen des prédictions quadratiques sur les données de démonstration.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

### Schéma explicatif
1. Entrée : Données brutes
2. Apprentissage : Tests de coupure conditionnelle (si/sinon)
3. Sortie : Valeur constante par morceaux (Feuilles)

### Contextes d'utilisation
- **Cas d'usage :** Grilles de tarification et d'évaluation.
- **Métriques :** MAE, RMSE.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom & Famille :** Arbre de décision pour la régression / *Decision Tree Regressor*. Famille des **arbres de décision** (apprentissage supervisé). Prédire une cible continue en partitionnant récursivement l'espace des caractéristiques.
- **Intuition & Étapes :** L'algorithme teste toutes les divisions possibles sur chaque variable pour trouver celle qui réduit au maximum la variance à l'intérieur des sous-groupes formés.
  1. *Entrée :* Caractéristiques $X$ et cible continue $y$.
  2. *Apprentissage :* Recherche séquentielle des seuils minimisant l'erreur quadratique (MSE).
  3. *Sortie :* Valeur constante prédite dans chaque feuille terminale.
- **Formule / Critère mathématique :**
  $$MSE = \frac{1}{N_m} \sum_{i \in R_m} (y_i - \bar{y}_{R_m})^2$$
  - *Symboles :* $R_m$ représente la région locale, $N_m$ le nombre d'observations dans cette région, et $\bar{y}_{R_m}$ la moyenne locale de la cible. *Paramètres :* seuils de coupure et moyennes des feuilles. *Hyperparamètres :* `max_depth` (profondeur maximale), `min_samples_split`.
- **Avantages & Limites :**
  - *Avantages :* (1) Très intuitif, visuel et entièrement explicable, (2) Gère les relations non linéaires sans normalisation.
  - *Limites :* (1) Très instable (sensible au bruit dans les données), (2) Sujet au surapprentissage si la profondeur n'est pas limitée.

### Schéma explicatif
- **Type de schéma :** Arbre de décision et partitionnement de l'espace.
- **Description du principe :**

```python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom & Famille :** Gradient Boosting pour la régression / *HistGradientBoostingRegressor*. Famille du **Boosting** (apprentissage supervisé). Construire un modèle robuste en combinant séquentiellement des arbres correcteurs.
- **Intuition & Étapes :** Les arbres s'enchaînent en cascade ; chaque nouvel arbre est entraîné spécifiquement pour corriger les erreurs résiduelles (les résidus) laissées par le modèle précédent.
  1. *Entrée :* Données $X$ et $y$.
  2. *Apprentissage :* Calcul des résidus du modèle actuel et ajustement d'un nouvel arbre sur ces résidus.
  3. *Sortie :* Somme pondérée des prédictions successives.
- **Formule / Critère mathématique :**
  $$F_m(X) = F_{m-1}(X) + \eta \cdot h_m(X)$$
  - *Symboles :* $F_m$ est le modèle cumulé à l'étape $m$, $h_m$ l'arbre correcteur, $\eta$ le taux d'apprentissage (*learning rate*). *Hyperparamètres :* `max_iter`, `learning_rate`, `max_depth`.
- **Avantages & Limites :**
  - *Avantages :* (1) Performances prédictives de pointe sur les données tabulaires, (2) Capture des relations complexes très finement.
  - *Limites :* (1) Entraînement séquentiel non parallélisable, (2) Risque de surapprentissage si le taux d'apprentissage ou le nombre d'itérations est mal réglé.

### Schéma explicatif
- **Type de schéma :** Diagramme de flux séquentiel (Correction des résidus).
- **Description du principe :**

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom & Famille :** K plus proches voisins / *K-Nearest Neighbors (KNN)*. Famille de la **classification / régression** (apprentissage supervisé)[cite: 1]. L'objectif est de prédire la classe ou la valeur d'une observation en se basant sur la majorité de ses $K$ voisins les plus proches dans l'espace des caractéristiques[cite: 1].
- **Intuition et étapes :** L'algorithme calcule la distance entre un point test et tous les points d'entraînement mémorisés, sélectionne les $K$ plus proches, et attribue la classe majoritaire par vote (ou la moyenne pour une régression).
- **Formule mathématique :**
  $$d(x, x') = \sqrt{\sum_{j=1}^{p} (x_j - x'_j)^2}$$
  - *Symboles :* $d(x, x')$ est la distance euclidienne entre deux observations, $p$ le nombre de variables, et $K$ l'hyperparamètre représentant le nombre de voisins pris en compte[cite: 1].
- **Avantages et limites :**
  - *Avantages :* Algorithme intuitif, non paramétrique, sans phase d'entraînement explicite (*lazy learning*).
  - *Limites :* Très lent en phase de prédiction sur de grands datasets et extrêmement sensible au fléau de la dimensionnalité ainsi qu'à l'échelle des variables[cite: 1].

### Schéma explicatif

```python
from sklearn.neighbors import KNeighborsClassifier
modele=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom & Famille :** Naïve Bayes Gaussien / *Gaussian Naive Bayes*. Famille de la **classification probabiliste** (apprentissage supervisé)[cite: 1]. L'objectif est d'estimer la probabilité d'appartenance d'une observation à différentes classes en appliquant le théorème de Bayes.
- **Intuition et étapes :** L'algorithme pose l'hypothèse simplificatrice (« naïve ») que toutes les variables sont indépendantes entre elles sachant la classe. Il calcule la probabilité a priori et les densités de probabilité gaussiennes de chaque attribut pour déterminer la classe maximale.
- **Formule mathématique :**
  $$P(y \mid X) \propto P(y) \prod_{j=1}^{p} P(X_j \mid y)$$
  - *Symboles :* $P(y \mid X)$ est la probabilité a posteriori, $P(y)$ la probabilité a priori de la classe, et $P(X_j \mid y)$ la densité de probabilité gaussienne de la variable $j$ (dont les moyennes et variances sont les paramètres appris)[cite: 1].
- **Avantages et limites :**
  - *Avantages :* Extrêmement rapide, performant en grande dimension et nécessite peu de données d'apprentissage.
  - *Limites :* L'hypothèse d'indépendance stricte entre les variables est rarement vérifiée dans la réalité[cite: 1].



```python
from sklearn.naive_bayes import GaussianNB
modele=GaussianNB().fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

## 14. Gradient Boosting — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom & Famille :** Gradient Boosting pour la classification / *HistGradientBoostingClassifier*. Famille du **Boosting / Ensembliste** (apprentissage supervisé). L'objectif est de prédire des classes en combinant séquentiellement des arbres de décision correcteurs.
- **Intuition et étapes :** Les arbres sont construits en cascade. Chaque nouvel arbre est entraîné pour prédire les erreurs résiduelles (gradients de la perte logistique) des arbres précédents afin d'affiner la frontière de décision globale.
- **Formule mathématique :**
  $$\mathcal{L}(y, f(X)) = - \sum \left[ y \log(p) + (1-y)\log(1-p) \right]$$
  - *Symboles :* $\mathcal{L}$ est la perte logistique (*log-loss*), $p = \sigma(f(X))$ la probabilité estimée par la somme des arbres, et le taux d'apprentissage $\eta$ l'hyperparamètre de pondération.
- **Avantages et limites :**
  - *Avantages :* Précision prédictive parmi les plus élevées sur données tabulaires et gestion native des valeurs manquantes.
  - *Limites :* Entraînement séquentiel plus lent et risque de surapprentissage si le nombre d'itérations est excessif.




```python
from sklearn.ensemble import HistGradientBoostingClassifier
modele=HistGradientBoostingClassifier(max_iter=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

## 15. AdaBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom & Famille :** Adaptive Boosting / *AdaBoost Classifier*. Famille du **Boosting / Ensembliste** (apprentissage supervisé). L'objectif est d'améliorer un classifieur simple en lui associant séquentiellement d'autres modèles correcteurs.
- **Intuition et étapes :** L'algorithme ré-associe des poids à chaque observation à chaque étape : les points mal classés voient leur poids augmenter, forçant l'arbre suivant à se concentrer en priorité sur ces cas difficiles.
- **Formule mathématique :**
  $$\alpha_m = \frac{1}{2} \ln \left( \frac{1 - e_m}{e_m} \right)$$
  - *Symboles :* $\alpha_m$ est le coefficient de pondération du classifieur $m$ dans le vote final, et $e_m$ son taux d'erreur pondéré sur l'échantillon d'apprentissage.
- **Avantages et limites :**
  - *Avantages :* Simplicité conceptuelle et mise en œuvre directe avec peu d'hyperparamètres à régler.
  - *Limites :* Très sensible au bruit et aux valeurs aberrantes (*outliers*) qui reçoivent des pénalités excessives.



```python
from sklearn.ensemble import AdaBoostClassifier
modele=AdaBoostClassifier(n_estimators=70,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```
